# use of matplotlib graph....

```python
#---use of matplotlib graph---#
 ## plot team wise total runs in bar charts

import matplotlib.pyplot as plt                                       
df.groupby('Team')['Runs'].sum().plot(kind='bar',color='pink')         
plt.title('Team-wise total Runs')###set charts title 
plt.xlabel('Team')##team name on x -axis
plt.ylabel('Runs')## runs on y- axis 
plt.show() ## show the garph
```

## plot pie chart for wickets distribution
## 0 wickets  = 40.8%  , 1 wickets  = 32.7%means most times bowlers take 0-1 wickets

```python
df['Wickets'].value_counts().plot(kind='pie',autopct='%1.1f%%',color=['red','lightgreen','brown'])
plt.title('Wickets Distribution')  ##set title......
plt.ylabel('')      # hide y - label for clean pie chart......
plt.show()  ## show graph.....
```

### plot histogram to runs distribution
# most bowlers conceded 25 - 35 runs bell shape graph

```python
df['Runs'].plot(kind='hist', bins=20, color='coral', edgecolor='black')
plt.title('Runs distribution')
plt.xlabel('Runs')
plt.ylabel('Frequency')
plt.show()
```
